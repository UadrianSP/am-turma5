# Pepeline Rondom Forest[Census]

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
random_forest_census = RandomForestClassifier(criterion='entropy', min_samples_leaf=1, min_samples_split=5, n_estimators=100)
random_forest_census.fit(X_credit_treinamento, y_credit_treinamento)
```

```python
rf_prediction = random_forest_census.predict(X_credit_teste)
rf_prediction
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
precisao = accuracy_score(y_credit_teste, rf_prediction)
print(f"Acurácia: {precisao * 100:.2f}%")
print("\nRelatório de Classificação:")
print(classification_report(y_credit_teste, rf_prediction))
```

```python
from yellowbrick.classifier import ConfusionMatrix

rf_cm = ConfusionMatrix(random_forest_census)
rf_cm.fit(X_credit_treinamento, y_credit_treinamento)
rf_cm.score(X_credit_teste, y_credit_teste)
rf_cm.poof()
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import export_graphviz
```

```python
arvore_census = DecisionTreeClassifier(criterion='entropy', min_samples_leaf=1, min_samples_split=5)
arvore_census.fit(X_credit_treinamento, y_credit_treinamento)
```

```python
DecisionTreeClassifier(
    ccp_alpha=0.0,
    class_weight=None,
    criterion='entropy',
    max_depth=None,
    max_features=None,
    max_leaf_nodes=None,
    min_impurity_decrease=0.0,
    min_samples_leaf=1,
    min_samples_split=2,
    min_weight_fraction_leaf=0.0,
    random_state=0,
    splitter='best'
)
```

```python
previsoes_arvore = arvore_census.predict(X_credit_teste)
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
acuracia_arvore = accuracy_score(y_credit_teste, previsoes_arvore)
print(f"Acurácia da Árvore: {acuracia_arvore * 100:.2f}%")

print("\nRelatório de Classificação:")
print(classification_report(y_credit_teste, previsoes_arvore))
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt


arvore_census_curta = DecisionTreeClassifier(criterion='entropy', max_depth=3, random_state=0)
arvore_census_curta.fit(X_credit_treinamento, y_credit_treinamento)

plt.figure(figsize=(20, 10))
plot_tree(arvore_census_curta, filled=True, rounded=True, class_names=True)
plt.show()
```
