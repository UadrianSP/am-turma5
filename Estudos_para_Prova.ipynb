```python
import pickle
with open('/content/drive/MyDrive/am-t5/census.pkl', 'rb') as f:
    census = pickle.load(f)
```

```python
X_census_train, y_census_train, X_census_test, y_census_test = census

X_census_train.shape, y_census_train.shape
```

```python
X_census_test.shape, y_census_test.shape
```

# Montagem do Neurônio Especialista

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

modelo = GaussianNB()
modelo.fit(X_census_train, y_census_train)

previsoes = modelo.predict(X_census_test)
print("Acurácia:", accuracy_score(y_census_test, previsoes))
print(classification_report(y_census_test, previsoes))
```

```python
neuronio_nb = GaussianNB()
neuronio_nb.fit(X_census_train, y_census_train)
```

```python
sinteiro_census = neuronio_nb.predict(X_census_test)
```

## Testes de Acurácia do Modelo

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
accuracy_score(y_census_test, sinteiro_census)
```

```python
from yellowbrick.classifier import ConfusionMatrix
cm = ConfusionMatrix(GaussianNB())
cm.fit(X_census_train, y_census_train)
cm.score(X_census_test, y_census_test)
cm.show()
```

```python
print(classification_report(y_census_test, sinteiro_census))
```
